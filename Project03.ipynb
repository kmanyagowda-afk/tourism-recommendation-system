````python
import pandas as pd

# Re-load csat_df and create final_df1 to ensure it's defined
# This assumes 'eCommerce_Customer_support_data.csv' is available in the environment.
csat_df = pd.read_csv('eCommerce_Customer_support_data.csv')
final_df1 = csat_df.copy()

# Display missing values before imputation
print("Missing values before imputation:")
print(final_df1.isnull().sum())
print("-" * 30)

# Impute Customer Remarks with 'No Review'
final_df1['Customer Remarks'].fillna("No Review", inplace=True);

# Impute Item_price with median
final_df1['Item_price'].fillna(final_df1['Item_price'].median(), inplace=True);

# Impute Customer_City and Product_category with 'Unknown'
final_df1['Customer_City'].fillna("Unknown", inplace=True);
final_df1['Product_category'].fillna("Unknown", inplace=True);

# Convert 'order_date_time' to datetime and then impute with mode
# First, convert to datetime, errors='coerce' will turn unparseable dates into NaT
final_df1['order_date_time'] = pd.to_datetime(final_df1['order_date_time'], format='%d/%m/%Y %H:%M', errors='coerce');
# Then, impute NaT values with the mode of the column
if not final_df1['order_date_time'].mode().empty:
    final_df1['order_date_time'].fillna(final_df1['order_date_time'].mode()[0], inplace=True);
else:
    # If mode is empty, fill with a more appropriate placeholder or strategy if needed
    final_df1['order_date_time'].fillna(pd.NaT, inplace=True); # Keeping NaT for now if no clear mode

# For 'connected_handling_time', the user mentioned creating a binary flag earlier, not direct imputation here.

# Display missing values after imputation
print("Missing values after imputation:")
print(final_df1.isnull().sum());
````

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Function to treat outliers using IQR
def treat_outliers_iqr(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Capping outliers
        df[col] = np.where(df[col] < lower_bound, lower_bound, df[col])
        df[col] = np.where(df[col] > upper_bound, upper_bound, df[col])

    return df

# Apply to numerical columns
df = treat_outliers_iqr(df, num_cols)

print("Outliers treated successfully")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

**IQR Method** (Interquartile Range Method)

Used to detect and cap extreme values.

Outliers defined as values outside:
Q1 − 1.5×IQR and Q3 + 1.5×IQR

Preferred because it works well for skewed distributions.

**Capping** (Winsorization)

Instead of deleting rows, extreme values were capped to boundary limits.

Prevents data loss while reducing the impact of extreme values.

These techniques improve model performance by reducing the effect of extreme values without removing important data
